## Step 1: Look at the type
Before analyzing anything, check what kind of data you're dealing with.
Numbers that are really categories (like `HighBP` being 0/1) get treated
differently than numbers that are actual measurements (like `BMI`).

````python
print(df.dtypes)

print(df["GenHlth"].unique())   # pick a column — what values does it actually take?
````

The yes/no columns and coded groups are categories, not measurements. For
example, averaging `HighBP`, `Age`, or `Education` would not be meaningful.
Converting these columns to `category` tells pandas to treat their values as
groups. `BMI`, `MentHlth`, and `PhysHlth` stay numeric because their values
represent measurements or counts.

````python
category_vars = [
    "Diabetes_binary", "HighBP", "HighChol", "CholCheck", "Smoker", "Stroke",
    "HeartDiseaseorAttack", "PhysActivity", "Fruits", "Veggies",
    "HvyAlcoholConsump", "AnyHealthcare", "NoDocbcCost", "DiffWalk", "Sex",
    "GenHlth", "Age", "Education", "Income",
]

for col in category_vars:
    df[col] = df[col].astype("category")

print(df.dtypes)
````

## Step 2: Look at the target alone
How many people have diabetes vs. not? Fill in the blanks below.

````python
counts = df["___"].value_counts().sort_index()

plt.bar(["No Diabetes", "Diabetes"], counts.values)
plt.title("Diabetes Status")
plt.ylabel("Number of people")
plt.show()
````

## Step 3: Compare a categorical column to the target
Pick one: `HighBP`, `HighChol`, `Smoker`, `Stroke`, `HeartDiseaseorAttack`,
`PhysActivity`, `DiffWalk`, `Age`, `GenHlth`, `Education`, or `Income`.

Because the target is categorical, use a normalized cross-tabulation to find
the percentage with diabetes within each category.

````python
column = "___"

# Each row adds to 100%; column 1 is the percentage with diabetes
proportions = pd.crosstab(
    df[column],
    df["Diabetes_binary"],
    normalize="index"
) * 100

print(proportions)

rate = proportions[1]

plt.bar(rate.index.astype(str), rate.values)
plt.title(f"Diabetes rate by {column}")
plt.xlabel(column)
plt.ylabel("% with diabetes")
plt.show()
````

---
## Step 4: Compare a numeric column to the target
Pick one: `BMI`, `MentHlth`, or `PhysHlth`.

````python
column = "____"
group0 = df.loc[df["Diabetes_binary"] == 0, column]
group1 = df.loc[df["Diabetes_binary"] == 1, column]

plt.boxplot([group0, group1], tick_labels=["No Diabetes", "Diabetes"])
plt.title(f"{column} by diabetes status")
plt.ylabel(column)
plt.show()
````

## Step 5: Look at one numeric column's spread
