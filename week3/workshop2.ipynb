````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("diabetes_binary_health_indicators_BRFSS2015.csv")
df.head()
````

---
## Step 1: Look at the type (setup — same as last time)

````python
category_vars = [
    "Diabetes_binary", "HighBP", "HighChol", "CholCheck", "Smoker", "Stroke",
    "HeartDiseaseorAttack", "PhysActivity", "Fruits", "Veggies",
    "HvyAlcoholConsump", "AnyHealthcare", "NoDocbcCost", "DiffWalk", "Sex",
    "GenHlth", "Age", "Education", "Income",
]

for col in category_vars:
    df[col] = df[col].astype("category")

print(df.dtypes)
````

---
## Step 2: Line chart
A line chart shows a trend — here, how the diabetes rate changes as `Age`
group increases. (Age is already ordered 1 → 13, youngest → oldest.)

````python
rate_by_age = df.groupby("Age")["Diabetes_binary"].value_counts(normalize=True).unstack()[1] * 100

plt.plot(rate_by_age.index.astype(int), rate_by_age.values, marker=".")
plt.title("Diabetes rate by age group")
plt.xlabel("Age group (1=youngest...13=oldest)")
plt.ylabel("% with diabetes")
plt.show()
````

---
## Step 3: Horizontal bar chart
Same idea as a regular bar chart, just sideways — useful when category
labels are long. Try `Education` or `Income`.

````python
column = "____"
rate = df.groupby(column)["Diabetes_binary"].value_counts(normalize=True).unstack()[1] * 100

plt.barh(rate.index.astype(str), rate.values)
plt.title(f"Diabetes rate by {column}")
plt.xlabel("% with diabetes")
plt.show()
````

---
## Step 4: Scatter plot
Shows the relationship between two numeric columns. Pick two different
ones: `BMI`, `MentHlth`, `PhysHlth`.

````python
x_column = "____"
y_column = "____"

plt.scatter(df[x_column], df[y_column], s=10, alpha=0.2)
plt.title(f"{x_column} vs. {y_column}")
plt.xlabel(x_column)
plt.ylabel(y_column)
plt.show()
````

---
## Step 5: Pie chart
Shows how a whole breaks into parts. Try `GenHlth` or `Education`. (Pie
charts get messy with more than ~5 slices — pick a column with few
categories.)

````python
column = "___"
counts = df[column].value_counts().sort_index()

plt.pie(counts.values, labels=counts.index.astype(str), autopct="%1.0f%%")
plt.title(f"Share of respondents by {column}")
plt.show()
````

---
## Step 6: Stacked bar chart
Shows two things at once: the size of each group *and* how diabetes splits
within it. Try `Sex` or `HighChol`.

````python
column = "Sex"
proportions = df.groupby(column)["Diabetes_binary"].value_counts(normalize=True).unstack()

proportions.plot(kind="bar", stacked=True)
plt.title(f"Diabetes proportion within each {column} group")
plt.ylabel("Proportion")
plt.xticks(rotation=0)
plt.show()
````
