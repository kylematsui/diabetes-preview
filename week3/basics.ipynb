# matplotlib basics — diabetes dataset
Same basic charts as before, now using real columns instead of made-up lists.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("diabetes_binary_health_indicators_BRFSS2015.csv")
df.head()
```

## Line plot
BMI for the first 30 people in the file.

```python
first_30 = df["BMI"].head(30)

plt.plot(first_30)
plt.title("BMI for the first 30 people")
plt.xlabel("Person #")
plt.ylabel("BMI")
plt.show()
```

## Bar chart
How many people have diabetes vs. not.

```python
counts = df["Diabetes_binary"].value_counts()

plt.bar(["No Diabetes", "Diabetes"], counts)
plt.title("How many people have diabetes?")
plt.ylabel("Number of people")
plt.show()
```

## Scatter plot
BMI vs. Age, first 200 people.

```python
sample = df.head(200)

plt.scatter(sample["Age"], sample["BMI"])
plt.title("Age group vs. BMI")
plt.xlabel("Age group (1=youngest...13=oldest)")
plt.ylabel("BMI")
plt.show()
```

## Histogram
Distribution of BMI across everyone.

```python
plt.hist(df["BMI"], bins=30)
plt.title("Distribution of BMI")
plt.xlabel("BMI")
plt.ylabel("Number of people")
plt.show()
```

## Pie chart
Share of people at each general-health rating.

```python
genhlth_counts = df["GenHlth"].value_counts().sort_index()
labels = ["Excellent", "Very Good", "Good", "Fair", "Poor"]

plt.pie(genhlth_counts, labels=labels, autopct="%1.0f%%")
plt.title("Self-rated general health")
plt.show()
```
